# EDA

### TRANSIT PASS-UPS

```python
df_pass = pd.read_csv("C:/UofM/01. Current Courses/04. Database Concepts/final project/Data/Transit_Pass-ups_20260928.csv")

#Prints number of rows each feature is missing data for (are NA)
print("Shape : " , df_pass.shape)
for col in df_pass.columns:
    print(f"\033[1mColumn Name: {col}\033[0m")
    print(f"Null values: {df_pass[col].isna().sum()}")
    print(f"Unique values: {df_pass[col].nunique()}")
#Prints total number of rows which have at least one missing value
print(f"Total unusable rows due to missing data: {df_pass.isna().any(axis=1).sum()}")
```

### ON-TIME PERFORMANCE

```python
df_perf = pd.read_csv("C:/UofM/01. Current Courses/04. Database Concepts/final project/Data/Transit_On-Time_Performance_Route_Aggregate_Data_20260928.csv",low_memory=False)

print("Shape : " , df_perf.shape)

#Prints number of rows each feature is missing data for (are NA)
for col in df_perf.columns:
    print(f"\033[1mColumn Name: {col}\033[0m")
    print(f"Null values: {df_perf[col].isna().sum()}")
    print(f"Unique values: {df_perf[col].nunique()}")
    print(f"Unique values: {df_perf[col].unique()}")
#Prints total number of rows which have at least one missing value
print(f"Total unusable rows due to missing data: {df_perf.isna().any(axis=1).sum()}")
```
